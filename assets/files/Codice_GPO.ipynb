# 1. Domanda e offerta
1) Calcola prezzo e quantità di equilibrio del mercato e rappresentali in un grafico insieme alle
curve di domanda e offerta.
2) Calcola l’elasticità della domanda di mercato per ogni punto e rappresentatela in un grafico.
Quando la domanda è elastica e quando è rigida?




```python
#Importo le librerire che mi servono
import numpy as np
import matplotlib.pyplot as plt

# Dichiaro le variabili

Qe = 600
Prezzi = np.linspace(1, 299, 100) #Prendi 100 valori da 1 a 299
P = 180
Qd = 1500 - 5*Prezzi
Qs = -300 + 5*Prezzi
C = 1200


PVariato = 200
VariazioneDomanda = (1500 - 5*200) - 600
VariazionePrezzo = 200 -180

# VariazionePercDomanda = VariazioneDomanda /Qe *100
# VariazionePercPrezzo = VariazionePrezzo /P *100

VariazionePercDomanda = VariazioneDomanda / Qd * 100
VariazionePercPrezzo = VariazionePrezzo / Prezzi * 100

Elasticita = abs(VariazionePercDomanda / VariazionePercPrezzo) #Questa formula prende il modulo dell'elasticità usando la sua formula
#Cl =50*Q + Q*Q

# Calcolo prezzo equilibrio e quantità equilibrio del mercato e li rappresento in un grafico insieme a domanda e offerta
# Faccio sistema tra Qd e Qs (fatto su carta)

plt.plot(Qe , P, "ro") #Equilibrio
plt.plot(Qs,Prezzi) #Offerta
plt.plot(Qd,Prezzi) #Domanda

plt.xlabel("Quantità (licenze)")
plt.ylabel("Prezzo (€)")
plt.title("Equilibrio di mercato: domanda e offerta")

plt.legend(["Equilibrio", "Offerta", "Domanda"])
plt.grid() # Fa vedere le griglie nel grafico
plt.show()


# Disegno il grafico dell'elasticità
plt.plot(Prezzi, Elasticita, label="Elasticità della domanda") # Do un nome a questa linea nella legenda (label)

plt.axhline(1, color="red", linestyle="--", #Crea la linea tratteggiata dell'elastictà unitaria
            label="Elasticità unitaria")

plt.xlabel("Prezzo (€)") #Aggiunge un nome all'ase X
plt.ylabel("Elasticità") #Aggiunge un nome all'ase Y
plt.title("Elasticità della domanda di mercato") # Da un titolo al grafico

plt.legend() #Aggiunge la legenda
plt.grid()
plt.show()

# Sopra i 150 euro la domanda è elastica
# Sotto i 150 è rigida
# Quando è a 150 è precisa a 1

```
